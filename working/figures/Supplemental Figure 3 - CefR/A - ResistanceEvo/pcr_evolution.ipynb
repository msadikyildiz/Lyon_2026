```
import os
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "reproduce.py").is_file())
os.chdir(ROOT)
OUTPUT_DIR = ROOT / "working/analysis/stats-rework/out/trajectories"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["mathtext.fontset"] = "custom"
plt.rcParams["mathtext.rm"] = "Times New Roman"


# Load the dataset
file_path = ROOT / 'working/figures/Supplemental Figure 3 - CefR/A - ResistanceEvo/pcr_evolution.xlsx'
pcrEVO = pd.read_excel(file_path)

# Calculate mean and standard deviation of Concentration grouped by Day
grouped = pcrEVO.groupby('Day')['Concentration']
mean_concentration = grouped.mean()
sd_concentration = grouped.std()

# Prepare data for plotting
days = mean_concentration.index

# Plot the data
plt.figure(figsize=(12, 10))  # Set figure size
plt.errorbar(days, mean_concentration, yerr=sd_concentration, fmt='o-', capsize=5, label='Mean Concentration ± SD', color='purple', linewidth=5, alpha=1)

# Add axis labels
plt.xlabel('Day', fontsize=40)
plt.ylabel('Maximum Allowable \nConcentration (μg/mL)', fontsize=40)


# Customize the y-axis to log scale with specific ticks
plt.yscale('log', base=2)
yticks = [0.05, 0.1, 0.2, 0.4, 0.8, 1.6, 3.2, 6.4, 12.8, 25.6, 51.2]
plt.yticks(yticks, labels=[str(tick) for tick in yticks], fontsize=35)

# Customize the x-axis ticks and labels to show only odd days
odd_days = days[days % 2 != 0]  # Filter odd days
plt.xticks(odd_days, labels=odd_days.astype(int), fontsize=35)

# Make axis lines and tick marks bolder, remove top and right spines
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
for spine in ['left', 'bottom']:
    ax.spines[spine].set_linewidth(4)
ax.tick_params(axis='both', which='major', length=12, width=3)
ax.tick_params(axis='both', which='minor', length=8, width=2)

# Add legend
# plt.legend(fontsize=35)

# Adjust spacing
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

# Save the plot
plt.savefig(OUTPUT_DIR / 'pcrEVO_plot.png', dpi=600, bbox_inches='tight')

# Show the plot
plt.show()
```
